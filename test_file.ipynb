```python
box_score = LE.BoxScore(*Timberwolves.iloc[0]) #the asterisk allows the pandas series to iterate over each feature (column)
#can only be used in a function
```

```python
box_score2 = LE.BoxScore(*Spurs.iloc[0]) #the asterisk allows the pandas series to iterate over each feature (column)
#can only be used in a function
```

```python
box_score.check_consistency()
```

```python
box_score.points_scored
```

```python
possessions = box_score.possessions()
print(possessions)
# Print the estimated number of possessions for the given box score.
# 105.44
```

```python
possessions2 = box_score2.possessions()
print(possessions2)
# Print the estimated number of possessions for the given box score.
# 99.64
```

```python
Stats = LE.TeamStats("Timberwolves", box_score, box_score2)
```

```python
Stats2 = LE.TeamStats("Spurs", box_score2, box_score)
```

```python
print(Stats.offensive_rating())
print(Stats.defensive_rating())
print(Stats.net_rating())
eFG, TOV, ORB, FT = Stats.four_factors()
```
